# 03 · The Telegram Bot API — no AI yet!

An **API** is a way for programs to talk to each other over the internet. We send a request to a URL, and get JSON back.

## Step 0 — Create your bot (on your phone)
1. In Telegram, search for **@BotFather** (blue tick ✅)
2. Send `/newbot`
3. Give it a name, e.g. `Mei Ling's Kopi Uncle`
4. Give it a username ending in `bot`, e.g. `meiling_kopi_uncle_bot`
5. BotFather replies with a **token** like `123456:ABC-DEF...` — keep it secret, like a password!
6. Paste it into your `.env` file: `TELEGRAM_BOT_TOKEN=123456:ABC-DEF...`

```python
import os
from getpass import getpass

import requests
from dotenv import load_dotenv

load_dotenv()  # reads your .env file

TOKEN = os.environ.get("TELEGRAM_BOT_TOKEN") or getpass("Paste your bot token: ")
BASE = f"https://api.telegram.org/bot{TOKEN}"

print("Token loaded:", TOKEN[:6] + "...")   # never print the whole token!
```

## 1. getMe — who am I?

```python
response = requests.get(f"{BASE}/getMe")
data = response.json()

print(data)
```

```python
print(data["result"]["first_name"])
print("@" + data["result"]["username"])
```

## 2. getUpdates — any new messages?
👉 **Send any message to your bot on Telegram first**, then run the next cell.

```python
response = requests.get(f"{BASE}/getUpdates")
updates = response.json()["result"]

print(f"{len(updates)} message(s) waiting")

for u in updates:
    message = u.get("message", {})
    # stickers and photos have no "text" - .get() stops them crashing our code
    print(message.get("from", {}).get("first_name"), ":", message.get("text", "(not text)"))
```

```python
# save the chat id of the latest message - we need it to reply
# (IndexError? Your bot has no messages yet - message it and re-run the cell above)
latest = updates[-1]
chat_id = latest["message"]["chat"]["id"]
name = latest["message"]["from"]["first_name"]

print(chat_id, name)
```

## 3. sendMessage — talk back!
Sending data uses **POST** instead of GET.

```python
response = requests.post(f"{BASE}/sendMessage", json={
    "chat_id": chat_id,
    "text": f"Hello {name}! Uncle can talk already 😎",
})

print(response.json()["ok"])
```

## 4. More fun API calls

```python
# roll a dice in the chat 🎲 (try 🎯 🏀 ⚽ 🎰 too)
response = requests.post(f"{BASE}/sendDice", json={"chat_id": chat_id, "emoji": "🎲"})

print("You rolled:", response.json()["result"]["dice"]["value"])
```

```python
# send a location: Lau Pa Sat hawker centre
requests.post(f"{BASE}/sendLocation", json={
    "chat_id": chat_id,
    "latitude": 1.2806,
    "longitude": 103.8504,
})
```
