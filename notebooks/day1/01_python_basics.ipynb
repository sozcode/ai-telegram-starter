## 1. print and variables

```python
# print shows something on the screen
print("Hello from Kopi Uncle!")
```

```python
# a variable is a labelled box that stores a value
bot_name = "Kopi Uncle"
kopi_price = 1.40
cups_sold = 3

print(bot_name)
print(kopi_price * cups_sold)
```

```python
# every value has a type
print(type(bot_name))     # str   = text
print(type(kopi_price))   # float = decimal number
print(type(cups_sold))    # int   = whole number
```

## 2. Strings — what the user types

```python
message = "  Uncle, one KOPI-O please  "

# clean it up so it's easier to check
print(message.strip())          # remove spaces at both ends
print(message.lower())          # make everything lowercase
print(message.strip().lower())  # do both
```

```python
# `in` checks if some text is inside other text
clean = message.strip().lower()

print("kopi" in clean)
print("milo" in clean)
```

```python
# f-strings put variables inside text
name = "Mei Ling"
total = kopi_price * 2

print(f"Hi {name}, 2 kopi is ${total:.2f}")   # :.2f = 2 decimal places
```

## 3. if / elif / else — making decisions

```python
text = "hello uncle"

if "hello" in text:
    print("Hello! What you want to drink?")
elif "kopi" in text:
    print("One kopi coming up!")
else:
    print("Aiyo, Uncle don't understand.")
```

**Try it:** change `text` above to `"one kopi please"` and `"bye"`, then run the cell again.

## 4. Functions — Uncle's first brain
A function is a reusable recipe: give it **input**, get back **output**.

```python
def reply(text):
    text = text.lower().strip()

    if "hello" in text:
        return "Hello! What you want to drink?"
    elif "kopi" in text:
        return f"One kopi coming up! That's ${kopi_price:.2f}."
    else:
        return "Aiyo, Uncle don't understand."


print(reply("Hello!"))
print(reply("KOPI please"))
print(reply("what's the time?"))
```

```python
# a function can take more than one input
def greet(name, drink):
    return f"Morning {name}! Your usual {drink}?"


print(greet("Ah Seng", "teh-o"))
```
