# 02 · Lists, Dictionaries & JSON — how the internet passes data around

Every Telegram message your bot receives is just **dictionaries inside dictionaries**. Learn these and you can read anything an API sends you.

## 1. Lists — an ordered collection

```python
drinks = ["kopi", "kopi-o", "teh", "milo"]

print(drinks[0])      # first item (counting starts at 0!)
print(drinks[-1])     # last item
print(len(drinks))    # how many
```

```python
# add an item
drinks.append("teh-c")
print(drinks)
```

```python
# a for-loop goes through each item one by one
for drink in drinks:
    print(f"We have {drink}")
```

## 2. Dictionaries — look things up by name

```python
prices = {
    "kopi": 1.40,
    "kopi-o": 1.20,
    "teh": 1.40,
    "milo": 1.80,
}

print(prices["milo"])
```

```python
# .get() won't crash if the key is missing
print(prices.get("bubble tea"))          # None
print(prices.get("bubble tea", "Got no"))
```

```python
# add or change a value
prices["teh-c"] = 1.50
prices["milo"] = 2.00
print(prices)
```

```python
# loop through keys and values together
for drink, price in prices.items():
    print(f"{drink}: ${price:.2f}")
```

## 3. Lists + dictionaries together

```python
order = ["kopi", "kopi", "milo"]

total = 0
for drink in order:
    total = total + prices[drink]

print(f"Total: ${total:.2f}")
```

## 4. A real Telegram message
This is exactly what Telegram sends your bot when someone types *"one kopi-o peng"*.

```python
update = {
    "update_id": 81723401,
    "message": {
        "message_id": 52,
        "from": {"id": 123456789, "is_bot": False, "first_name": "Mei Ling", "language_code": "en"},
        "chat": {"id": 123456789, "first_name": "Mei Ling", "type": "private"},
        "date": 1763190000,
        "text": "one kopi-o peng"
    }
}

# dig in, one level at a time
print(update["message"]["text"])
print(update["message"]["from"]["first_name"])
print(update["message"]["chat"]["id"])
```

## 5. JSON — dictionaries as text
APIs send data as **JSON text**. Python turns it into dictionaries with `json.loads`, and back with `json.dumps`.

```python
import json

raw = '{"ok": true, "result": {"first_name": "Kopi Uncle", "username": "my_kopi_uncle_bot"}}'

data = json.loads(raw)     # text -> dictionary
print(type(data))
print(data["result"]["username"])
```

```python
# dictionary -> pretty JSON text
print(json.dumps(update, indent=2))
```

## 6. A list of updates
When you ask Telegram for new messages, you get back a **list** of updates.

```python
updates = [
    {"update_id": 1, "message": {"from": {"first_name": "Mei Ling"}, "chat": {"id": 111}, "text": "hello"}},
    {"update_id": 2, "message": {"from": {"first_name": "Ahmad"}, "chat": {"id": 222}, "text": "one milo"}},
    {"update_id": 3, "message": {"from": {"first_name": "Priya"}, "chat": {"id": 333}, "text": "teh-c kosong"}},
]

for u in updates:
    name = u["message"]["from"]["first_name"]
    text = u["message"]["text"]
    print(f"{name} says: {text}")
```
