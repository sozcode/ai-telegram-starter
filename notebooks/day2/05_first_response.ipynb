# 05 · Talking to the AI 🤖

Yesterday Uncle's brain was a pile of `if` statements. Today we give him a real brain: **OpenAI's Responses API**.

The idea is simple: we send some **input**, the AI sends back a **response**.

```python
import os
from getpass import getpass

from dotenv import load_dotenv
from openai import OpenAI

if not os.path.exists("main.py"):   # make sure we're in the project folder
    os.chdir("../..")

load_dotenv()
if not os.environ.get("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass("Paste your OpenAI API key: ")

client = OpenAI()
MODEL = "gpt-6-luna"
print("Ready!")
```

## 1. Your first response

```python
response = client.responses.create(
    model=MODEL,
    input="Uncle, what drink you recommend for a hot afternoon?",
)

print(response.output_text)
```

`response.output_text` is the reply as text. Run the cell again — the AI doesn't answer the same way twice!

## 2. Instructions = personality
`instructions` tell the AI **who it is** and **how to behave**.

```python
response = client.responses.create(
    model=MODEL,
    instructions="You are a friendly kopitiam uncle in Singapore. Speak light Singlish. Keep replies short.",
    input="Uncle, what drink you recommend for a hot afternoon?",
)

print(response.output_text)
```

```python
grumpy = "You are a grumpy but lovable kopitiam uncle. Complain a lot, then help anyway. Max 2 lines."

response = client.responses.create(model=MODEL, instructions=grumpy, input="Uncle, kopi or teh better?")
print(response.output_text)
```

## 3. Thinking speed
`gpt-6-luna` can *reason* (think) before answering. More thinking = smarter but slower. For a chat bot, **low** is usually best.

```python
import time

for effort in ["low", "medium"]:
    start = time.time()
    response = client.responses.create(
        model=MODEL,
        reasoning={"effort": effort},
        input="Kopi or teh is better? Why? One line.",
    )
    print(f"[{effort}] {time.time() - start:.1f}s -> {response.output_text}")
```

## 4. The problem: it makes things up 😬

```python
uncle = "You are a friendly kopitiam uncle in Singapore. Speak light Singlish. Keep replies short."

response = client.responses.create(model=MODEL, instructions=uncle, input="How much is one kopi-o peng and two kaya toast?")
print(response.output_text)
```

Uncle just **invented** a price. He has never seen our menu!

**Next notebook:** we give him **tools** — so he can look up the real menu.

## 5. Swap the brain
Remember `reply()` from Day 1? Here it is with an AI brain.

```python
def reply(text):
    response = client.responses.create(model=MODEL, instructions=uncle, input=text)
    return response.output_text


print(reply("can I have a kopi but less sweet?"))
```
