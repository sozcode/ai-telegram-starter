# 05 · Talking to the AI 🤖

Yesterday's bot could only follow `if` rules. Today we build **SG Data Bot** — an AI assistant that knows about life in Singapore and can work with real data.

We use **OpenAI's Responses API**: we send some **input**, the AI sends back a **response**.

```python
import os
from getpass import getpass

from dotenv import load_dotenv
from openai import OpenAI

if not os.path.exists("main.py"):   # make sure we're in the project folder
    os.chdir("../..")

load_dotenv()
if not os.environ.get("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass("Paste your OpenAI API key: ")

client = OpenAI()
MODEL = "gpt-6-luna"
print("Ready!")
```

## 1. Your first response

```python
response = client.responses.create(
    model=MODEL,
    input="Suggest one fun thing to do in Singapore this weekend.",
)

print(response.output_text)
```

`response.output_text` is the reply as text. Run the cell again — the AI doesn't answer the same way twice!

## 2. Instructions = personality
`instructions` tell the AI **who it is** and **how to behave**.

```python
bot = "You are SG Data Bot, a helpful assistant for life in Singapore. Keep replies short."
```

```python
response = client.responses.create(
    model=MODEL,
    instructions=bot,
    input="Suggest one fun thing to do in Singapore this weekend.",
)

print(response.output_text)
```

```python
auntie = "You are a kaypoh Singaporean auntie who has an opinion on everything. Speak Singlish. Max 2 lines."

response = client.responses.create(model=MODEL, instructions=auntie, input="Should I buy a resale flat or a BTO?")
print(response.output_text)
```

## 3. Thinking speed
`gpt-6-luna` can *reason* (think) before answering. More thinking = smarter but slower. For a chat bot, **low** is usually best.

```python
import time

for effort in ["low", "medium"]:
    start = time.time()
    response = client.responses.create(
        model=MODEL,
        reasoning={"effort": effort},
        input="BTO or resale flat - which is better for a young couple? One line.",
    )
    print(f"[{effort}] {time.time() - start:.1f}s -> {response.output_text}")
```

## 4. The problem: it doesn't know *right now* 😬

```python
for question in ["Is it raining in Tampines right now?", "What's the median 4-room HDB resale price in Tampines this year?"]:
    response = client.responses.create(model=MODEL, instructions=bot, input=question)
    print(question)
    print("->", response.output_text, "\n")
```

The AI only knows what it learned during training. It can't see today's weather or the latest HDB sales — so it either says *"I don't know"* or **guesses**.

**Next notebooks:** we give it **tools** to fetch live data and analyse real files.

## 5. Swap the brain
Remember `reply()` from Day 1? Here it is with an AI brain.

```python
def reply(text):
    response = client.responses.create(model=MODEL, instructions=bot, input=text)
    return response.output_text


print(reply("what's a good hawker centre near Tiong Bahru?"))
```
