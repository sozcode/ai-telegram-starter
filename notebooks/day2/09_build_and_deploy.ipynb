## Step 2 — Test it here
After editing `main.py`, **restart the kernel** (⟳ at the top) so Python loads your changes.

````python
import asyncio
import os

if not os.path.exists("main.py"):   # make sure we're in the project folder
    os.chdir("../..")

from main import ask_uncle
````

````python
# ask_uncle is the same function Telegram will trigger.
# (asyncio.to_thread lets the bot's normal code run inside a notebook)
for text in ["hello", "what do you have?", "2 kopi-o peng, how much?"]:
    answer = await asyncio.to_thread(ask_uncle, 1, text)
    print("YOU:", text)
    print("BOT:", answer, "\n")
````

## Step 4 — Deploy to Vercel 🌍
Yesterday you put `day1_bot.py` online. Now switch Vercel to your AI bot:

1. Open **`pyproject.toml`** and change the last line to:
   ```
   entrypoint = "main:app"
   ```
2. **vercel.com → your project → Settings → Environment Variables** → add `OPENAI_API_KEY`, `SUPABASE_URL` and `SUPABASE_SECRET_KEY` (you added `TELEGRAM_BOT_TOKEN` and `WEBHOOK_SECRET` yesterday).
3. **Source Control** tab → **Commit** → **Sync Changes**. Vercel redeploys by itself.

> 🆕 Didn't deploy yesterday? Follow the *Put your bot online* steps at the end of notebook 04, but use `main:app`.

## Step 6 — Tell Telegram where your bot lives 🔔
Telegram already sends messages to your Vercel address from yesterday. Run this cell if you're deploying for the first time, or if you ran `python main.py` (which switches the bot back to polling).

````python
from dotenv import load_dotenv

load_dotenv()
TOKEN = os.environ.get("TELEGRAM_BOT_TOKEN")
WEBHOOK_SECRET = os.environ.get("WEBHOOK_SECRET")   # the same password you put in Vercel

response = requests.post(f"https://api.telegram.org/bot{TOKEN}/setWebhook", json={
    "url": APP_URL + "/webhook",
    "secret_token": WEBHOOK_SECRET,
})
response.json()
````

`'ok': True`? 🎉 Stop `python main.py`, close your laptop — your bot still replies!

> 💡 Running `python main.py` again switches your bot back to polling. Re-run the cell above when you're done.
