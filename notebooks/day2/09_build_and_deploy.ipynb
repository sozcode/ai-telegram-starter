## Step 2 — Test it here
After editing files in `bot/`, **restart the kernel** so Python loads your changes.

````python
import asyncio
import os

from dotenv import load_dotenv

if not os.path.exists("data"):   # make sure we're in the project folder
    os.chdir("../..")
load_dotenv()

from bot.handler import brain, make_reply

print("Brain:", brain())
````

````python
# make_reply is the same function Telegram will trigger.
# (asyncio.to_thread lets the bot's normal code run inside a notebook)
for text in ["hello", "what do you have?", "2 kopi-o peng, how much?"]:
    answer = await asyncio.to_thread(make_reply, 1, text, "Tester")
    print("YOU:", text)
    print("BOT:", answer, "\n")
````

## Step 4 — Deploy to Vercel 🌍
1. **Source Control** tab → type a message → **Commit** → **Sync Changes** (pushes to your GitHub).
2. **vercel.com** → **Add New → Project** → import your repo → **Deploy**.
3. **Settings → Environment Variables** → add `TELEGRAM_BOT_TOKEN`, `OPENAI_API_KEY`, `SUPABASE_URL`, `SUPABASE_SECRET_KEY` and `WEBHOOK_SECRET` (any random text).
4. **Deployments → ⋯ → Redeploy** (new variables only apply after a redeploy).
5. Stop `run_polling.py`, then open **`https://YOUR-APP.vercel.app/setup`** once.

Your bot now runs 24/7 — close your laptop and it still replies! 🎉
