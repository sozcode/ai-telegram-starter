## Step 4 — Deploy to Vercel 🌍
Yesterday you put `day1_bot.py` online. Now switch Vercel to your AI bot:

1. Open **`pyproject.toml`** and change the last line to:
   ```
   entrypoint = "main:app"
   ```
2. **vercel.com → your project → Settings → Environment Variables** → add `OPENAI_API_KEY` and `OPENAI_CONVERSATION_ID` (copy both from your `.env`).
3. **Source Control** tab → **Commit** → **Sync Changes**. Vercel redeploys by itself.

> 🆕 Didn't deploy yesterday? Follow the *Put your bot online* steps at the end of notebook 04, but use `main:app`.

## Step 6 — Tell Telegram where your bot lives 🔔
Telegram already sends messages to your Vercel address from yesterday. Run this cell if you're deploying for the first time, or if you ran `python main.py` (which switches the bot back to polling).

````python
from dotenv import load_dotenv

load_dotenv()
TOKEN = os.environ.get("TELEGRAM_BOT_TOKEN")
WEBHOOK_SECRET = os.environ.get("WEBHOOK_SECRET")   # the same password you put in Vercel

response = requests.post(f"https://api.telegram.org/bot{TOKEN}/setWebhook", json={
    "url": APP_URL + "/webhook",
    "secret_token": WEBHOOK_SECRET,
})
response.json()
````

`'ok': True`? 🎉 Close your laptop — your bot still replies!
